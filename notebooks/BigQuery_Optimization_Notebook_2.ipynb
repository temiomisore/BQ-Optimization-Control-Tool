### 🟢 Step 4: Execute the Two-Person Approval Workflow (§8)
For Class 3 structural rebuilds (`demo_ecommerce.iot_sensor_telemetry`), two distinct sign-offs are strictly required:
1. **Sign-off 1 (Asset Owner)**: Marks partial approval; state remains `PENDING_REVIEW`.
2. **Sign-off 2 (Platform / FinOps)**: Completes authorization; state transitions to `APPROVED`.


````python
# Sign-off 1: Asset Owner signs off
signoff_1_sql = f"""
UPDATE `{PROJECT_ID}.optimizer_ops.change_sets`
SET 
  approvals = ARRAY_CONCAT(COALESCE(approvals, []), [
    STRUCT('alice-owner@company.com' AS principal, CURRENT_TIMESTAMP() AS `at`, 'owner_approver' AS role)
  ]),
  state_history = ARRAY_CONCAT(state_history, [
    STRUCT('PENDING_REVIEW' AS state, CURRENT_TIMESTAMP() AS `at`, 'alice-owner@company.com' AS actor, 'Sign-off 1/2 complete (Asset Owner)' AS note)
  ])
WHERE target_table = 'iot_sensor_telemetry';
"""
client.query(signoff_1_sql).result()
print("✅ Sign-off 1/2 recorded! State remains PENDING_REVIEW (execution locked).")

# Sign-off 2: Platform / FinOps signs off
signoff_2_sql = f"""
UPDATE `{PROJECT_ID}.optimizer_ops.change_sets`
SET 
  state = 'APPROVED',
  approvals = ARRAY_CONCAT(COALESCE(approvals, []), [
    STRUCT('finops-lead@company.com' AS principal, CURRENT_TIMESTAMP() AS `at`, 'platform_approver' AS role)
  ]),
  state_history = ARRAY_CONCAT(state_history, [
    STRUCT('APPROVED' AS state, CURRENT_TIMESTAMP() AS `at`, 'finops-lead@company.com' AS actor, 'Sign-off 2/2 complete (Platform Approver)' AS note)
  ])
WHERE target_table = 'iot_sensor_telemetry';
"""
client.query(signoff_2_sql).result()
print("✅ Sign-off 2/2 recorded! State transitioned to APPROVED (execution unlocked).")
````

````python
query_approvals = f"""
SELECT 
  c.target_table,
  c.state,
  app.principal AS approver,
  app.role AS approval_role,
  app.`at` AS approved_at
FROM `{PROJECT_ID}.optimizer_ops.change_sets` c,
UNNEST(c.approvals) AS app
WHERE c.target_table = 'iot_sensor_telemetry';
"""
df_approvals = client.query(query_approvals).to_dataframe()
display(df_approvals)
````

## 🔬 4. Deep Dive: Technical Evidence, Schema DDL & Mechanics

### 1️⃣ Dynamic Workload-Capped Honest Scoring (§6.2)
- **Problem**: Native tools evaluate rules in isolation; a recommendation may claim $8,000/mo on an $800/mo table.
- **Remediation**: Savings claim is dynamically capped:
  $$\text{Capped} = \min(\text{Claim}, \text{Table Spend} \times 0.68)$$
- Confidence haircut: $d_{\text{summation}} = 0.50$ halves confidence for multi-stage overlap.

### 2️⃣ Multi-Tiered Ownership Attribution (§8, §9.1)
- **Hierarchy**: `Table Label` → `Git Codeowners` → `Top Active 30d Writer` → `Dataset Admin`.

### 3️⃣ Two-Person Approval Governance (§8)
- Requires distinct sign-offs from both **Asset Owner** and **Platform/FinOps** before any structural table rebuild.

### 4️⃣ Time-Travel Window Tuning (`C1-04`)
- Underlying SQL: `ALTER SCHEMA SET OPTIONS (max_time_travel_hours = 48);`
- Eliminates 60% of physical storage churn on high-update tables.

### 5️⃣ Unenforced Primary & Foreign Keys (`C1-10`)
- Underlying SQL: `ALTER TABLE ... ADD PRIMARY KEY (id) NOT ENFORCED;`
- Enables BigQuery query optimizer join elimination and reordering.

### 6️⃣ Search Indexes with Cost Watchdogs (`C2-02`)
- Accelerates point lookups while capping index maintenance storage costs.

### 7️⃣ Partition Granularity Tuning (`C3-03`)
- Rebuilds micro-partitioned daily tables into monthly partitions to prevent nearing the 10,000 partition ceiling.
